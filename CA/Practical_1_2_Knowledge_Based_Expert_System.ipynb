```python
facts = {
    "soil": input("Is the soil dry? (yes/no): ").lower(),
    "leaves": input("Are the leaves yellow? (yes/no): ").lower(),
    "sunlight": input("Is the plant getting enough sunlight? (yes/no): ").lower()
}

recommendations = []

if facts["soil"] == "yes":
    recommendations.append("Water the plant.")

if facts["leaves"] == "yes":
    recommendations.append("Check the plant for overwatering or nutrient deficiency.")

if facts["sunlight"] == "no":
    recommendations.append("Move the plant to a place with sufficient sunlight.")

if not recommendations:
    recommendations.append("The plant appears healthy. Continue regular care.")

print("\nPlant Care Recommendations:")
for recommendation in recommendations:
    print("-", recommendation)
```
