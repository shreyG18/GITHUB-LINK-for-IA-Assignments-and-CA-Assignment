```python
class VacuumEnvironment:
    def __init__(self):
        self.locations = {
            "A": "Dirty",
            "B": "Dirty"
        }
        self.agent_location = "A"

    def percept(self):
        return self.agent_location, self.locations[self.agent_location]

    def clean(self):
        self.locations[self.agent_location] = "Clean"


class SimpleReflexVacuumAgent:
    def action(self, location, status):
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Move Right"
        else:
            return "Move Left"


environment = VacuumEnvironment()
agent = SimpleReflexVacuumAgent()

for step in range(6):
    location, status = environment.percept()
    action = agent.action(location, status)

    print("Step:", step + 1)
    print("Location:", location)
    print("Status:", status)
    print("Action:", action)
    print()

    if action == "Suck":
        environment.clean()
    elif action == "Move Right":
        environment.agent_location = "B"
    elif action == "Move Left":
        environment.agent_location = "A"
```
